## Using ``console.table()``

Displays dictionaries and list in a table

```typescript



const studentName = "Alex"
const favoriteSubjects = ["Math", "Science", "Coding"]

// 2. Print a regular message using console.log
console.log("Student Name:", studentName)
console.log("First favorite subject is:", favoriteSubjects[0])

// 3. Print the whole list as a clean grid using console.table
console.table(favoriteSubjects)
```
