# 04 -  Nested Example
when a if statement is inside a if statement

```typescript
const height = Number(prompt("Welcome, how tall are you (cm)?: "));
if (height >= 120){
    const age = Number(prompt("How old are you?: "));
    if (age > 18){
        console.log("Approved");
    } else{
        console.log("Denied");
    }
}else{
    console.log("Denied");
}
```

# 05 - Multiple If - Statement

```typescript
const age = 80;
if (age < 18 && age >= 0) {
    console.log("Kid");
} if (age >= 18 && age < 60) {
    console.log("Adult");
} if (age >= 60) {
    console.log("Senior");
}
```

# 06 - else If

```typescript
const age = 80;
if (age < 18 && age >= 0) {
    console.log("Kid");
} else if (age >= 18 && age < 60) {
    console.log("Adult");
} else if (age >= 60) {
    console.log("Senior");
}
```
