# 02 - Concatenate string and type conversion

```typescript
console.log("Hello, your name is " + "Ethan " + "Lin"  );
```

```typescript
const num_Char = prompt("What is your name?: ").length;
console.log("Your name is has " + `${num_Char}` + " character");
```

```typescript
const num_Char = prompt("What is your name?: ").length;
const temp = String(num_Char);
console.log("Your name is " + `${temp} ` + "characters long.");
```

# 03 - Mathematical Operations

```typescript
console.log(3 + 5);

console.log(7 - 4);

console.log(7 - 10);

console.log(2 * 3); // multiplication

console.log(2.0 * 3); // JavaScript treats all numbers as floats internally

console.log(6.3 / 3); // division always gives a decimal number

console.log(Math.floor(7.0 / 3)); // floor division (rounds down)

console.log(Math.floor(7 / 3)); // force round down to nearest integer

console.log(2 ** 3); // this is the power ex: 2^3

console.log(9 % 3); // finds the remainder
```

$bmi = Weight(kg) / Height(meter)^2$
