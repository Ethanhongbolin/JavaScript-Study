# 03 - Variable
**Rules**:
- Can contain letters``AbCdf``, numbers``12345``, underscores``_``, and dollar sign``$``.
- Must start with a letter, underscore or dollar sign
so it CAN NOT start with number.
- Case Sensitive: ``totalPrice``, ``TotalPrice`` and ``TOTALPRICE`` are entirely different variebles.
- Cannot be JavaScript keywords: you cannot name a variable after a reserved words like ``const``,``let``,``if``,``while``, or ``function``.

**Best Practice**
- Use descriptive names: Use ``age`` instead of a.
- Use camelCase: Unlike Python's ``snake_case``, the absolute standard in JavaScript is to start with a lowercase letter and capitalize the first letter of each added word (e.g., ``totalPrice``, ``isLoggedIn``).
- Use UPPER_CASE for fixed constants: If a variable is a hardcoded value that will never change throughout the lifetime of the app, use all caps with underscores (e.g., ``const MAX_LOGIN_ATTEMPTS = 5;``).
- Avoid single-letter names: Except for index counters in loops (like i).
- Never capital the first letter 
- Use semi colons``;`` at the end of each line exept if statements or functions or loops
